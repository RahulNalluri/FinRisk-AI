```python
X_cc = creditcard.drop('Class', axis=1)
y_cc = creditcard['Class']
```

```python
from sklearn.model_selection import train_test_split

X_train_cc, X_test_cc, y_train_cc, y_test_cc = train_test_split(
    X_cc,
    y_cc,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model_cc = RandomForestClassifier(n_estimators=100, random_state=42)

model_cc.fit(X_train_cc, y_train_cc)
```

```python
y_pred_cc = model_cc.predict(X_test_cc)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test_cc, y_pred_cc)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test_cc, y_pred_cc))
```

```python
import os
import joblib

os.makedirs("../models", exist_ok=True)

joblib.dump(model_cc, "../models/creditcard_model.pkl")
```

```python
scam.fillna(scam.median(numeric_only=True), inplace=True)
```

```python
scam = pd.get_dummies(scam, drop_first=True)
```

```python
X_scam = scam.drop('is_fraudulent', axis=1)
y_scam = scam['is_fraudulent']
```

```python
from sklearn.model_selection import train_test_split

X_train_scam, X_test_scam, y_train_scam, y_test_scam = train_test_split(
    X_scam,
    y_scam,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model_scam = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model_scam.fit(X_train_scam, y_train_scam)
```

```python
y_pred_scam = model_scam.predict(X_test_scam)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test_scam, y_pred_scam))
print(classification_report(y_test_scam, y_pred_scam))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test_scam, y_pred_scam)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')

plt.title("Scam Detection Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

model_loan = LogisticRegression(max_iter=1000)

model_loan.fit(X_train_loan, y_train_loan)
```

```python
y_pred_loan = model_loan.predict(X_test_loan)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test_loan, y_pred_loan))
print(classification_report(y_test_loan, y_pred_loan))
```
